```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Dataframe5').getOrCreate()
```

```python
spark
```

```python
df_pyspark = spark.read.csv('tips.csv', header=True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```

```python
df_pyspark.columns
```

```python
#  handling categorical features
from pyspark.ml.feature import StringIndexer
```

```python
indexer=StringIndexer(inputCol='sex', outputCol='sex_indexed')
df_r = indexer.fit(df_pyspark).transform(df_pyspark)
df_r.show()
```

```python
indexer=StringIndexer(inputCols=['smoker','day','time'], outputCols=['smoker_indexed','day_indexed','time_indexed'])
df_r = indexer.fit(df_r).transform(df_r)
df_r.show()
```

```python
from pyspark.ml.feature import VectorAssembler
```

```python
featureassembler = VectorAssembler(inputCols = ['tip','sex_indexed','smoker_indexed','day_indexed','time_indexed'],outputCol = 'Independent Features')
output = featureassembler.transform(df_r)
```

```python
output.show()
```

```python
output.select('Independent Features').show()
```

```python
finalized_data = output.select('Independent Features','total_bill')
```

```python
finalized_data.show()
```

```python
from pyspark.ml.regression import LinearRegression

# train test split
train_data,test_data = finalized_data.randomSplit([0.75,0.25])
regressor = LinearRegression(featuresCol='Independent Features', labelCol='total_bill')
regressor = regressor.fit(train_data)
```

```python
regressor.coefficients
```

```python
regressor.intercept
```

```python
pred_results = regressor.evaluate(test_data)
```

```python
pred_results.predictions.show()
```

```python
# performance metrics
pred_results.r2, pred_results.meanAbsoluteError, pred_results.meanSquaredError
```
