```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('Dataframe2').getOrCreate()
```

```python
spark
```

```python
df_pyspark = spark.read.csv('test1.csv', header=True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
# drop the column
df_pyspark.drop('Name').show()
```

```python
df_pyspark.show()
```

```python
df_pyspark.na.drop().show()
```

```python
# any==how
df_pyspark.na.drop(how='any').show()
```

```python
# if how == any, drop the column with atleast one null value
# if how == all, drop the column with all the value as null
```

```python
# threshold
df_pyspark.na.drop(how='any', thresh=2).show()
# atleast 2 non-null values should be there
```

```python
# subset
df_pyspark.na.drop(how='any', subset=['Age']).show()
# drops the row if null value present in the mentioned column
```

```python
# filling the missing value
df_pyspark.na.fill('Missing value', 'Experience').show()
```

```python
df_pyspark.na.fill('Missing value', ['Name', 'Salary']).show()
```

```python
df_pyspark.na.fill(0, 'Salary').show()
```

```python
df_pyspark.show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols = ['Age', 'Experience', 'Salary'],
    outputCols = ["{}_imputed".format(c) for c in ['Age', 'Experience', 'Salary']]
    ).setStrategy('median')
```

```python
# add imputated columns in df
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
